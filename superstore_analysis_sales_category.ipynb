```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

conn = sqlite3.connect(r"sqlRSuperStore.db")
```

## Question 2: Sales, Sub-Categories

```python
query = '''
SELECT "Sub-Category", sum("Sales") AS total_sales
FROM "Sample - Superstore"
GROUP BY "Sub-Category"
ORDER BY total_sales DESC;
'''

sales = pd.read_sql_query(query, conn)
print(sales)
```

```python
plt.figure(figsize = (10, 6))
plt.barh(sales["Sub-Category"], sales["total_sales"])
plt.xlabel("Total Sales")
plt.title("Total Sales by Sub-Category")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig("sales_by_subcategory.png")
plt.show()
```
