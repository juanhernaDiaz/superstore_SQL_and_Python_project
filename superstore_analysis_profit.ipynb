```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

conn = sqlite3.connect(r"sqlRSuperStore.db")
```

## Question 1: Profit, Sub-Categories
Which sub-categories of products generate the most profits?

```python
query = '''
SELECT "Sub-Category", sum("Profit") AS total_profit
FROM "Sample - Superstore"
GROUP BY "Sub-Category"
ORDER BY total_profit DESC;
'''

profit = pd.read_sql_query(query, conn)
print(profit)
```

```python
plt.figure(figsize = (10, 6))
plt.barh(profit["Sub-Category"], profit["total_profit"])
plt.xlabel("Total Profit")
plt.title("Total Profit by Sub-Category")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig("profit_by_subcategory.png")
plt.show()
```

## Takeaway
Copiers, Phones, and Accessories are the most profitable sub-categories. Allocate resources from less profitable sub-categories into these sections and increase their availability in the store to continue driving sales and generating more profit. It is also advisable to check whether major discounting explains the losses by comparing average discount across sub-categories.

Supplies, Bookcases, and Tables are selling at a loss. Make an attempt to reduce manufacturing costs and/or slightly increase prices (increase prices is not advisable for tables). Alternatively, change their marketing to appeal to a different target audience or bundle these sub-categories with better profitable ones. For example, tables and supplies can be bundled with copiers.
