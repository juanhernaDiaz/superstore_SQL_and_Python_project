```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

conn = sqlite3.connect(r"sqlRSuperStore.db")
```

## Question 5: Sales, Region

```python
query = '''
SELECT
	"Discount",
	COUNT(*) AS "Sales"
FROM "Sample - Superstore"
GROUP BY "Discount"
ORDER BY "Discount" ASC;
'''

disc = pd.read_sql_query(query, conn)
print(disc)
```

```python
plt.figure(figsize=(10, 6))
plt.bar(disc["Discount"].astype(str), disc["Sales"])
plt.xlabel("Discount Level")
plt.ylabel("Number of Orders")
plt.title("Order Count by Discount Level")
plt.tight_layout()
plt.savefig("orders_by_discount.png")
plt.show()
```
